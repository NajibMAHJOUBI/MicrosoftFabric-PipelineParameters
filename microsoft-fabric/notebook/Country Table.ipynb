```python
import json
import pandas as pd
```

```python
name = "canada"
```

```python
with open(f"/lakehouse/default/Files/JSON/{name}.json", 'r') as f:
    data = json.load(f)
```

```python
#data
```

```python
pays = data[0]['name']['common']
capital = data[0]['capital'][0]
region = data[0]['region']
subregion = data[0]['subregion']
flags = data[0]['flags']['png']
```

```python
df = pd.DataFrame({'Pays': [pays],
'Capital': [capital],
'Region': [region],
'Subregion': [subregion],
'Flags': [flags]})
```

```python
df.to_csv(f"/lakehouse/default/Files/CSV/{name}.csv", index=False)
```
